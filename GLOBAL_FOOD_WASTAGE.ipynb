```python
import pandas as pd 
df = pd.read_csv("global_food_wastage_dataset(6).csv")
print(df)
print("First 5 Rows:")
print(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nColumns:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nStatistical Summary:")
print(df.describe())


print(df["Year"].min())
print(df["Year"].max())

print(df["Total Waste (Tons)"].min())
print(df["Economic Loss (Million $)"].min())

print(df["Avg Waste per Capita (Kg)"].min())
print(df["Population (Million)"].min())
print(df["Household Waste (%)"].min())
print(df["Household Waste (%)"].max())


print(df[df["Household Waste (%)"] > 100])
print(df[df["Household Waste (%)"] < 0])


total_waste = df["Total Waste (Tons)"].sum()
print("Total Food Waste:", total_waste, "Tons")

total_loss = df["Economic Loss (Million $)"].sum()
print("Total Economic Loss:", total_loss, "Million $")


avg_household = df["Household Waste (%)"].mean()
print("Average Household Waste:", avg_household, 

 country_waste = df.groupby("Country")["Total Waste (Tons)"].sum()
print(country_waste.sort_values(ascending=False).head(10))


category_waste = df.groupby("Food Category")["Total Waste (Tons)"].sum()
print(category_waste.sort_values(ascending=False))


year_waste = df.groupby("Year")["Total Waste (Tons)"].sum()
print(year_waste)

country_loss = df.groupby("Country")["Economic Loss (Million $)"].sum()
print(country_loss.sort_values(ascending=False).head(10))

per_capita = df.groupby("Country")["Avg Waste per Capita (Kg)"].mean()
print(per_capita.sort_values(ascending=False).head(10))

print(df.corr(numeric_only=True))
```
